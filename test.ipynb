```python
import cv2

img = cv2.imread("photo1.jpg")

resized_img = cv2.resize(
    img,
    None,
    fx=0.15,
    fy=0.15
)

cv2.imwrite("photo1resi.jpg", resized_img)
```

```python
## normal detection
import cv2
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

photo = cv2.imread("photo1resi.jpg")   # Original image
result = model(photo)

cv2.imshow("My Window", result[0].plot())
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
## Only face

import cv2
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

face_model = YOLO("yolov8m-face.pt")

photo = cv2.imread("photo1resi.jpg")   # Original image
result = face_model(photo)

cv2.imshow("My Window", result[0].plot())
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
##cat + face

import cv2
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

face_model = YOLO("yolov8m-face.pt")

photo = cv2.imread("demo.jpg")   # Original image
result = model(photo)
face_result = face_model(photo)

cat_person = result[0].plot()
cat_person_face = face_result[0].plot(img=cat_person)
cv2.imshow("my window", cat_person_face)

cv2.waitKey(5000)
cv2.destroyAllWindows()
```
